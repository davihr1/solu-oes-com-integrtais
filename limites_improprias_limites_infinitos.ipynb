```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.integrate import quad

# Configuração do estilo do seaborn
sns.set(style="whitegrid")

# Função integrando
def integrand3(x):
    return np.exp(-x**2)

# Cálculo da integral
result3, error3 = quad(integrand3, -np.inf, np.inf)
print(f"Integral de -∞ a ∞ de e^(-x^2) dx = {result3}")

# Dados para o gráfico
x3 = np.linspace(-3, 3, 400)
y3 = integrand3(x3)

# Gráfico
plt.figure(figsize=(10, 5))
sns.lineplot(x=x3, y=y3)
plt.fill_between(x3, y3, alpha=0.3)
plt.title('Integral de -∞ a ∞ de e^(-x^2) dx')
plt.xlabel('x')
plt.ylabel('e^(-x^2)')
plt.show()
```
