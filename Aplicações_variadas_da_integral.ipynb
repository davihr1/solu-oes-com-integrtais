# Aplicação variadas da integral Definida

# **9.1** Aplicações da Integral na Física

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

texto = "Trabalho, Energia Potencial Gravitacional, Momento de uma Força, Trajetoria de um projetil, Centro de Massa, Momento de Inercia, Calor, entropia, Eficiencia."

nuvem_palavras = WordCloud(width=800, height=400, background_color="white").generate(texto)
plt.figure(figsize=(10,5))
plt.imshow(nuvem_palavras, interpolation='bilinear')
plt.axis("off")
plt.show()
```
