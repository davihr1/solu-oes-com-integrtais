# 7 Integral definida e o Teorema Fundamental do Calculo

#7.1 Definição e propriedades da integral definida

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

#Define a vatiavél simbolica
x = sp.Symbol('x')

#Define a Função
f = x**2  # Corrected: f should be defined in terms of x

#Calcula a integral definida de f(x) no intevalo [0, 1]
integral_vale = sp.integrate(f, (x, 0, 1)) # Corrected: integrate syntax

#Converte a Função pra uma função numerica
f_num = sp.lambdify(x, f, 'numpy')

#Gera valores para x
x_vals = np.linspace(0, 1, 100)

#Calcula os valores de y
y_vals = f_num(x_vals)

#Plota a função
plt.figure(figsize=(8, 6))
plt.plot(x_vals, y_vals, label='$f(x) = x^2$')

#Preenche a área sob a curva entre 0 e 1
plt.fill_between(x_vals, y_vals, color='skyblue', alpha=0.3, label='Area sob a curva') # Corrected: standard color
plt.xlabel('X')
plt.ylabel('f(x)')
plt.title('Função $f(x) = x^2$ e Área sob a Curva') # Corrected: title typo and string closing
plt.legend()
plt.grid(True)
plt.show()

print('O valor da integral definida de f(x) no intervalo [0, 1] é: ', integral_vale.evalf()) # Corrected: SyntaxError, added comma and used integral_vale
```

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

def plot_integral(f, a, b):
  """
   plota a função f(x) e a area sob a curva no intervalo [a, b]

   Parametros:
   - f: função a ser integrada (função simbolica)
   - a: limite inferior do intervalo de integração
   - b: limite superior do intervalo de integração
  """

  #Define a variavel simbolica "X"
  x = sp.Symbol('x')

  #Calcula a integral definida de f(x) no intervalo [a, b]
  integral_vale = sp.integrate(f, (x, a, b))

  #Converte a função para uma função numerica
  f_num = sp.lambdify(x, f, 'numpy')

  #Gera valores para x
  x_vals = np.linspace(a, b, 100)

  #Calcula os valores de y
  y_vals = f_num(x_vals)

  #Plota a função
  plt.figure(figsize=(8, 6))
  plt.plot(x_vals, y_vals, label='$f(x)$')

  #Preenche a area sob a curva entre a e b
  plt.fill_between(x_vals, y_vals, color='skyblue', alpha=0.3, label='Area sob a curva')

  #Adiciona legendas e titulos
  plt.xlabel('X')
  plt.ylabel('f(x)')
  plt.title('Interpretação geometrica da integral definida')
  plt.legend()

  #Exibe o grafico
  plt.grid(True)
  plt.show()

  print("O valor da integral definida de f(x) no intervalo [", a, ", ", b, "] é:", integral_vale.evalf())


#Testa a função com a função f(x) = nl(x) e o intervalo [1, 2]
x = sp.Symbol('x')
f = sp.ln(x)
plot_integral(f, 1, 2)
print("Chamada da função plot_integral concluída.")
```
