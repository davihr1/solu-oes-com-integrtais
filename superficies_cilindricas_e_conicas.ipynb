## 13.1. Superfícies Cilíndricas: Definição e Características

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Definição dos parâmetros do cilindro
r = 2  # Raio do cilindro
h = 5  # Altura do cilindro

# Criação dos pontos para o gráfico
theta = np.linspace(0, 2*np.pi, 100)  # Ângulos variando de 0 a 2*pi
z = np.linspace(0, h, 100)  # Alturas variando de 0 a h
theta, z = np.meshgrid(theta, z)  # Criação da grade de pontos
x = r * np.cos(theta)  # Coordenada x
y = r * np.sin(theta)  # Coordenada y

# Criação do gráfico tridimensional
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Plotagem da superfície cilíndrica
ax.plot_surface(x, y, z, color='lightblue', alpha=0.8)

# Destaque do raio
ax.plot([0, r], [0, 0], [0, 0], color='red', linewidth=2, label='Raio')

# Destaque da geratriz
ax.plot([r, r], [0, 0], [0, h], color='green', linewidth=2, label='Geratriz')

# Destaque da diretriz
ax.plot([0, 0], [0, 0], [0, h], color='orange', linewidth=2, label='Diretriz')

# Configurações adicionais do gráfico
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Superfície Cilíndrica')

# Adiciona legenda
ax.legend()

# Exibição do gráfico
plt.show()
```

## 13.2. Superfícies Cônicas: Tipos e Propriedades

```python
import matplotlib.pyplot as plt
import numpy as np

# Parâmetros do cone
r = 3  # Raio da base
h = 4  # Altura
theta = np.linspace(0, 2*np.pi, 200)  # Ângulos para a base
z = np.linspace(0, h, 200)  # Alturas ao longo do cone

# Gerando pontos na superfície do cone
X = np.outer(r * np.cos(theta), np.sqrt(1 - (z/h)**2))
Y = np.outer(r * np.sin(theta), np.sqrt(1 - (z/h)**2))
Z = np.outer(np.ones_like(theta), z)

# Plotando o cone em 3D
fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='k')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Cone Circular Reto')
ax.set_box_aspect([1, 1, h/r])  # Aspecto ajustado para a proporção correta

plt.show()
```

## 13.3. Equações de Superfícies Cilíndricas em Coordenadas Cartesianas

```python
import matplotlib.pyplot as plt
import numpy as np

# Parâmetros do cilindro elíptico
a = 3  # Semi-eixo maior
b = 5  # Semi-eixo menor
h = 5  # Altura do cilindro
theta = np.linspace(0, 2 * np.pi, 200) # Ângulos para a parametrização
z = np.linspace(0, h, 200) # Altura ao longo do cilindro

# Criação da grade de coordenadas
Theta, Z = np.meshgrid(theta, z)

# Parametrização do cilindro
X = a * np.cos(Theta)
Y = b * np.sin(Theta)
Z = Z

# Plotando o cilindro elíptico em 3D
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Superfície do cilindro
ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='k', alpha=0.7)

# Legendas e título
ax.set_xlabel('X', fontsize=12)
ax.set_ylabel('Y', fontsize=12)
ax.set_zlabel('Z', fontsize=12)
ax.set_title('Cilindro Elíptico', fontsize=14)

# Aspecto da caixa
ax.set_box_aspect([a, b, h]) # Ajusta o aspecto da caixa para refletir as proporções do cilindro

# Adicionando uma legenda manualmente
h_patch = plt.Line2D([0], [0], linestyle="none", marker="|", markersize=10, markerfacecolor="g")
ax.legend([h_patch], ['Cilindro Elíptico'], loc='best')

plt.show()
```

## 13.4. Equações de Superfícies Cônicas em Coordenadas Cartesianas

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configurar estilo do Seaborn
sns.set(style='whitegrid')

# Parâmetros da elipse diretriz e altura do cone
a = 3  # Semi-eixo maior
b = 4  # Semi-eixo menor
h = 5  # Altura do cone (vértice ao plano da diretriz)

# Ângulos para a parametrização da elipse diretriz
theta = np.linspace(0, 2 * np.pi, 100)

# Coordenadas da elipse diretriz no plano z = 0
x_diretriz = a * np.cos(theta)
y_diretriz = b * np.sin(theta)
z_diretriz = np.zeros_like(theta)

# Coordenadas do vértice
x_vertice = 0
y_vertice = 0
z_vertice = h

# Coordenadas da superfície do cone
X = np.outer(x_diretriz, np.linspace(0, 1, 100))
Y = np.outer(y_diretriz, np.linspace(0, 1, 100))
Z = np.outer(z_diretriz, np.linspace(1, 0, 100)) * h

# Geratrizes (linhas do vértice até a diretriz)
geratrizes_x = np.vstack((np.zeros_like(x_diretriz), x_diretriz))
geratrizes_y = np.vstack((np.zeros_like(y_diretriz), y_diretriz))
geratrizes_z = np.vstack((np.ones_like(z_diretriz) * h, z_diretriz))

# Plotando o cone e seus componentes
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Superfície do cone
ax.plot_surface(X, Y, Z, alpha=0.5, cmap='viridis', edgecolor='k')

# Vértice do cone
ax.scatter([x_vertice], [y_vertice], [z_vertice], color='r', s=100, label='Vértice')

# Elipse diretriz
ax.plot(x_diretriz, y_diretriz, z_diretriz, color='b', label='Diretriz (Elipse)')

# Geratrizes
for i in range(len(theta)):
    ax.plot(geratrizes_x[:, i], geratrizes_y[:, i], geratrizes_z[:, i], color='g', alpha=0.6)

# Legendas e título
ax.set_xlabel('X', fontsize=12)
ax.set_ylabel('Y', fontsize=12)
ax.set_zlabel('Z', fontsize=12)
ax.set_title('Superfície Cônica com Geratriz, Diretriz e Vértice', fontsize=14)

# Ajuste da escala dos eixos
ax.set_box_aspect([a, b, h])

# Exibição da legenda
ax.legend()
plt.show()
```
