### **8.1. Princípio do Método de Integração por Partes**

$$\int udv = uv - \int vdu$$

---

* exemplo: calcular

$$\int x\cos(x)dx$$

* solução:

1. :

$$\int x\cos(x)dx = \int udv$$

2. :

$$\int udv = uv - \int vdu$$

3. :

$$\int udv = x\text{sen}(x) - \int \text{sen}(x)dx$$

4. :

$$\int udv = x\text{sen}(x) + \cos(x) + C$$

* Conclusão:

5. :

$$\int x\cos(x)dx = x\text{sen}(x) + \cos(x) + C$$

```python
import sympy as sp

x = sp.symbols('x')

# Define u and dv for the integral of x * cos(x) dx
u = x
dv_expr = sp.cos(x)

# Calculate du (derivative of u) and v (integral of dv_expr)
du_expr = sp.diff(u, x)
v_expr = sp.integrate(dv_expr, x)

# Apply the integration by parts formula: integral(u dv) = u*v - integral(v du)
integral = u * v_expr - sp.integrate(v_expr * du_expr, x)

print(integral)
```

# 8.2 Aplicação do metodo em Integrais Definidas

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.integrate import quad

# Definindo a função
def func(x):
  return x * np.sin(x)

# Calculando a Integral
integral_result, _ = quad(func, 0, np.pi)
print('O resultado da integral definida é: ', integral_result)

# Criando os valores x e y para plotagem
x_values = np.linspace(0, np.pi, 1000)
y_values = func(x_values)

# Plotando o Grafico
plt.figure(figsize=(10, 6))
plt.plot(x_values, y_values, label=r'$x\sin(x)$') # Usando raw string para LaTeX
plt.fill_between(x_values, y_values, alpha=0.3)
plt.xlabel('x')
plt.ylabel('y')
plt.title(r'Grafico de $x\sin(x)$') # Usando raw string para LaTeX
plt.legend()

# Estilizando com seaborn
sns.set()
plt.grid(True)
plt.show()
```

**Objetivo: Integrar a função**

$$x * ln(x)$$

**usando a integração por partes.**

**Aplicar a Regra LIATE:**
* L: Funções Logarítmicas (ln(x))
* I: Funções Inversas (Não se aplica neste caso)
* A: Função Algébrica (x)
* T: Função Trigonométrica (Não se aplica neste caso)
* E: Função Exponencial (Não se aplica neste caso)

**De acordo com a regra LIATE, a função logarítmica**

$$(ln(x))$$

**deve ser escolhida como**

$$u$$

**e a função algébrica**

$$(x)$$

**como**

$$dv$$

**Passo 2: Definir as Funções u e dv:**
* u(x) = ln(x)
* dv(x) = x dx

**Passo 3: Diferenciar u e Integrar dv:**
* $$du(x) = \frac{1}{x}dx$$
* $$v(x) = \int xdx = x^2 / 2$$

**Passo 4: Aplicar a Fórmula da Integração por Partes:**

$$\int u(x) * dv(x)dx = u(x) * v(x) - \int u'(x) * v(x)dx$$

**Substituindo as funções definidas:**

$$\int x * ln(x)dx = ln(x) * \frac{x^2}{2} - \int \frac{1}{x} * \frac{x^2}{2}dx$$

**Passo 5: Resolver a Integral Restante:**

$$\int x * ln(x)dx = x^2 * \frac{ln(x)}{2} - \int \frac{x}{2}dx$$

**Passo 6: Adicionar a Constante de Integração: (Resultado Final)**

$$\int x * ln(x)dx = x^2 * ln(x)/2 - \frac{x^2}{4} + C$$
