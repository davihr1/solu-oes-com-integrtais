## 14.1. Conceito de Superfícies de Revolução

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Parâmetros da curva senoidal
theta = np.linspace(0, 2*np.pi, 100)
x = np.sin(theta)
y = np.cos(theta)

# Parâmetros da superfície de revolução
u = np.linspace(0, 2*np.pi, 100)
v = np.linspace(-1, 1, 100)
u, v = np.meshgrid(u, v)

# Coordenadas da superfície de revolução
X = np.outer(np.cos(u), x)
Y = np.outer(np.sin(u), x)
Z = np.outer(np.ones(np.size(u)), y)

# Plotagem da superfície de revolução
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='k')

# Legendas e título
ax.set_xlabel('X', fontsize=12)
ax.set_ylabel('Y', fontsize=12)
ax.set_zlabel('Z', fontsize=12)
ax.set_title('Superfície de Revolução de uma Curva Senoidal', fontsize=14)

plt.show()
```

## 14.2. Método de Geração de Superfícies de Revolução


```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Função para gerar coordenadas da superfície de revolução de uma parábola
def generate_parabola_surface():
    # Definição da função geratriz: f(x) = x^2
    def f_parabola(u):
        return u**2

    # Parâmetros da superfície de revolução
    u_parabola = np.linspace(0, 1, 100)
    v = np.linspace(0, 2 * np.pi, 100)
    u_parabola, v = np.meshgrid(u_parabola, v)

    # Coordenadas da superfície de revolução para a parábola
    x_parabola = f_parabola(u_parabola)
    y_parabola = u_parabola * np.cos(v)
    z_parabola = u_parabola * np.sin(v)

    return x_parabola, y_parabola, z_parabola

# Função para gerar coordenadas da superfície de revolução de uma senoide
def generate_sine_surface():
    # Definição da função geratriz: f(x) = sin(x)
    def f_sine(u):
        return np.sin(u)

    # Parâmetros da superfície de revolução
    u_sine = np.linspace(0, np.pi, 100)
    v = np.linspace(0, 2 * np.pi, 100)
    u_sine, v = np.meshgrid(u_sine, v)

    # Coordenadas da superfície de revolução para a senoide
    x_sine = f_sine(u_sine)
    y_sine = u_sine * np.cos(v)
    z_sine = u_sine * np.sin(v)

    return x_sine, y_sine, z_sine

# Função para gerar coordenadas da superfície de revolução de uma elipse
def generate_ellipse_surface():
    # Definição dos parâmetros da elipse
    a = 3  # Semi-eixo maior
    b = 2  # Semi-eixo menor

    # Definição da função geratriz para a elipse
    def f_ellipse(u):
        return b * np.sqrt(1 - (u**2 / a**2))

    # Parâmetros da superfície de revolução
    u_ellipse = np.linspace(-a, a, 100)
    v = np.linspace(0, 2 * np.pi, 100)
    u_ellipse, v = np.meshgrid(u_ellipse, v)

    # Coordenadas da superfície de revolução para a elipse
    x_ellipse = u_ellipse
    y_ellipse = f_ellipse(u_ellipse) * np.cos(v)
    z_ellipse = f_ellipse(u_ellipse) * np.sin(v)

    return x_ellipse, y_ellipse, z_ellipse

# Gerar as coordenadas das superfícies de revolução
x_parabola, y_parabola, z_parabola = generate_parabola_surface()
x_sine, y_sine, z_sine = generate_sine_surface()
x_ellipse, y_ellipse, z_ellipse = generate_ellipse_surface()

# Plotar as superfícies
fig = plt.figure(figsize=(15, 5))

# Superfície de revolução da parábola
ax1 = fig.add_subplot(131, projection='3d')
ax1.plot_surface(x_parabola, y_parabola, z_parabola, cmap='viridis', edgecolor='k')
ax1.set_title('Superfície de Revolução: $y = x^2$', fontsize=14)
ax1.set_xlabel('X', fontsize=12)
ax1.set_ylabel('Y', fontsize=12)
ax1.set_zlabel('Z', fontsize=12)

# Superfície de revolução da senoide
ax2 = fig.add_subplot(132, projection='3d')
ax2.plot_surface(x_sine, y_sine, z_sine, cmap='viridis', edgecolor='k')
#ax2.set_title('Superfície de Revolução: $y = \\sin(x)$', fontsize=14)
ax2.set_title('Superfície de Revolução: $y = sen(x)$', fontsize=14)
ax2.set_xlabel('X', fontsize=12)
ax2.set_ylabel('Y', fontsize=12)
ax2.set_zlabel('Z', fontsize=12)

# Superfície de revolução da elipse (Completado para você)
ax3 = fig.add_subplot(133, projection='3d')
ax3.plot_surface(x_ellipse, y_ellipse, z_ellipse, cmap='viridis', edgecolor='k')
ax3.set_title('Superfície de Revolução: Elipse', fontsize=14)
ax3.set_xlabel('X', fontsize=12)
ax3.set_ylabel('Y', fontsize=12)
ax3.set_zlabel('Z', fontsize=12)

plt.tight_layout() # Ajuda a não encavalar os gráficos
plt.show()
```

## 14.3. Exemplos de Superfícies de Revolução em Coordenadas Cartesianas

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

theta = np.linspace(0, 2*np.pi, 100)
z = np.linspace(0, 1, 100)
theta, z = np.meshgrid(theta, z)
x = np.cos(theta) * z
y = np.sin(theta) * z

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(x, y, z, cmap='viridis', edgecolor='k')
ax.set_xlabel('X', fontsize=12)
ax.set_ylabel('Y', fontsize=12)
ax.set_zlabel('Z', fontsize=12)
ax.set_title('Superfície de Revolução: Cone', fontsize=14)
plt.show()
```
