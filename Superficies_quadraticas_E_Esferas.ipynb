# 12. Superfícies Quádricas e Esféricas

## 12.1. Superfícies Quádricas: Definição e Propriedades


```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Definindo a superfície do elipsoide
u = np.linspace(0, 2 * np.pi, 100)
v = np.linspace(0, np.pi, 100)
x = 3 * np.outer(np.cos(u), np.sin(v))
y = 2 * np.outer(np.sin(u), np.sin(v))
z = 1 * np.outer(np.ones(np.size(u)), np.cos(v))

# Gráfico do elipsoide
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(x, y, z, color='b', alpha=0.6)
ax.set_title('Elipsoide: $\\dfrac{x^2}{3^2} + \\dfrac{y^2}{2^2} + \\dfrac{z^2}{1^2} = 1$')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
plt.show()
```

# Exemplo: Esfera com Centro na Origem

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Definindo os ângulos para a superfície da esfera
u = np.linspace(0, 2 * np.pi, 100)
v = np.linspace(0, np.pi, 100)

# Equações paramétricas da esfera (raio r = 1)
r = 1
x = r * np.outer(np.cos(u), np.sin(v))
y = r * np.outer(np.sin(u), np.sin(v))
z = r * np.outer(np.ones(np.size(u)), np.cos(v))

# Configuração do Gráfico
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

# Plotando a superfície (adicionado edgecolor para mostrar a grade)
ax.plot_surface(x, y, z, color='steelblue', alpha=0.6, edgecolor='w', linewidth=0.3)

# Título e rótulos dos eixos
ax.set_title('Esfera: $x^2 + y^2 + z^2 = 1$')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')

# Ajustando os limites dos eixos para a esfera não ficar achatada
ax.set_xlim([-1, 1])
ax.set_ylim([-1, 1])
ax.set_zlim([-1, 1])

# Mostrando o gráfico
plt.show()
```

# Exemplo: Esfera com Centro em (h,k,l)


```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Definindo os ângulos para a superfície da esfera
u = np.linspace(0, 2 * np.pi, 100)
v = np.linspace(0, np.pi, 100)

# Centro da esfera (h, k, l) e raio r
h = 2
k = 3
l = 4
r = 1

# Equações paramétricas da esfera com centro deslocado
x = h + r * np.outer(np.cos(u), np.sin(v))
y = k + r * np.outer(np.sin(u), np.sin(v))
z = l + r * np.outer(np.ones(np.size(u)), np.cos(v))

# Configuração do Gráfico
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

# Plotando a superfície (utilizando uma cor avermelhada semelhante à da imagem)
ax.plot_surface(x, y, z, color='indianred', alpha=0.6, edgecolor='w', linewidth=0.3)

# Título e rótulos dos eixos
ax.set_title('Esfera: $(x - 2)^2 + (y - 3)^2 + (z - 4)^2 = 1$')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')

# Ajustando os limites dos eixos de acordo com o centro e o raio
ax.set_xlim([1, 3])
ax.set_ylim([2, 4])
ax.set_zlim([3, 5])

# Mostrando o gráfico
plt.show()
```

## 12.2. Superfícies Esféricas: Equações e Características

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Definindo a esfera
u = np.linspace(0, 2 * np.pi, 100)
v = np.linspace(0, np.pi, 100)
x = np.outer(np.cos(u), np.sin(v))
y = np.outer(np.sin(u), np.sin(v))
z = np.outer(np.ones(np.size(u)), np.cos(v))

# Gráfico da esfera
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(x, y, z, color='b', alpha=0.6)
ax.set_title('Esfera: $x^2 + y^2 + z^2 = 1$')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
plt.show()
```

# Exemplo: Cilindro Circular

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Definindo os parâmetros para a superfície do cilindro
theta = np.linspace(0, 2 * np.pi, 100)
z = np.linspace(-2, 2, 100)

# Criando a malha de coordenadas (meshgrid)
Theta, Z = np.meshgrid(theta, z)

# Equações paramétricas do cilindro circular (raio r = 1)
r = 1
X = r * np.cos(Theta)
Y = r * np.sin(Theta)

# Configuração do Gráfico
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

# Traçando a superfície (com cor semelhante à da imagem e grelha branca)
ax.plot_surface(X, Y, Z, color='steelblue', alpha=0.6, edgecolor='w', linewidth=0.3)

# Título e rótulos dos eixos
ax.set_title('Cilindro Circular: $x^2 + y^2 = 1$')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')

# Ajustando os limites dos eixos conforme a imagem
ax.set_xlim([-1, 1])
ax.set_ylim([-1, 1])
ax.set_zlim([-2, 2])

# Apresentando o gráfico
plt.show()
```

# Exemplo: Cone Circular

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Definindo os parâmetros para a superfície do cone
# r varia de 0 a 2 (limite do eixo z na imagem)
r = np.linspace(0, 2, 100)
# theta varia de 0 a 2*pi (círculo completo)
theta = np.linspace(0, 2 * np.pi, 100)

# Criando a malha de coordenadas (meshgrid)
R, Theta = np.meshgrid(r, theta)

# Equações paramétricas do cone (z = r para a parte superior)
X = R * np.cos(Theta)
Y = R * np.sin(Theta)
Z = R

# Configuração do Gráfico
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

# Traçando a superfície (com cor semelhante à da imagem e grelha branca)
ax.plot_surface(X, Y, Z, color='indianred', alpha=0.6, edgecolor='w', linewidth=0.3)

# Título e rótulos dos eixos
ax.set_title('Cone Circular: $z^2 = x^2 + y^2$')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')

# Ajustando os limites dos eixos conforme a imagem
ax.set_xlim([-2, 2])
ax.set_ylim([-2, 2])
ax.set_zlim([0, 2])

# Apresentando o gráfico
plt.show()
```

## 12.3. Superfícies Cilíndricas e Cônicas

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Definindo o cone
theta = np.linspace(0, 2 * np.pi, 100)
r = np.linspace(0, 2, 100)
theta, r = np.meshgrid(theta, r)
x = r * np.cos(theta)
y = r * np.sin(theta)
z = r

# Gráfico do cone
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(x, y, z, color='r', alpha=0.6)
ax.set_title('Cone Circular: $z^2 = x^2 + y^2$')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Definindo a curva geratriz do vaso
def vase_profile(y):
    if y < 1:
        return 0.5 + 0.5 * y
    elif y < 3:
        return 1.5 - 0.5 * (y - 1)
    else:
        return 1.0 + 0.2 * (y - 3)

# Valores de y
y = np.linspace(0, 5, 500)

# Aplicando a função para gerar o perfil
vase_r = np.vectorize(vase_profile)(y)

# Rotacionando a curva geratriz
theta = np.linspace(0, 2 * np.pi, 500)
theta, y = np.meshgrid(theta, y)
x = vase_r[:, None] * np.cos(theta)
z = vase_r[:, None] * np.sin(theta)

# Gráfico do vaso
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(x, y, z, color='skyblue', alpha=0.7)
ax.set_title('Design de um Vaso usando Superfície de Revolução')
ax.set_xlabel('X')
ax.set_ylabel('Altura (Y)')
ax.set_zlabel('Z')
plt.show()
```
